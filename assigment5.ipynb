```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Create graph with specified friendship connections
G5 = nx.Graph()
friendships = [
    ("Alice", "Bob"),
    ("Alice", "Charlie"),
    ("Bob", "David"),
    ("Charlie", "David"),
    ("David", "Eva"),
    ("Eva", "Frank"),
    ("Charlie", "Frank")
]
G5.add_edges_from(friendships)

# 3. Display total number of nodes and edges
num_nodes_5 = G5.number_of_nodes()
num_edges_5 = G5.number_of_edges()
print(f"Total Nodes: {num_nodes_5}")
print(f"Total Edges: {num_edges_5}")

# 4. Compute and print degree of each student
degrees_5 = dict(G5.degree())
print("\nDegree of each student:")
for student, deg in degrees_5.items():
    print(f"  {student}: {deg}")

# 5. Identify student(s) with highest number of friends
max_degree_5 = max(degrees_5.values())
top_students = [student for student, deg in degrees_5.items() if deg == max_degree_5]
print(f"\nStudent(s) with highest number of friends ({max_degree_5} friends): {', '.join(top_students)}")

# 6. Check direct friendship between Alice and Eva
has_direct_link = G5.has_edge("Alice", "Eva")
print(f"\nDirect friendship between Alice and Eva: {has_direct_link}")

# 7. Shortest path and length between Alice and Frank.
shortest_path_5 = nx.shortest_path(G5, source="Alice", target="Frank")
path_length_5 = nx.shortest_path_length(G5, source="Alice", target="Frank")
print(f"Shortest path between Alice and Frank: {' -> '.join(shortest_path_5)}")
print(f"Shortest path length: {path_length_5}")

# 2. Visualize the network graph
plt.figure(figsize=(8, 6))
pos_5 = nx.spring_layout(G5, seed=42)

nx.draw_networkx_nodes(G5, pos_5, node_color='skyblue', node_size=1200)
nx.draw_networkx_edges(G5, pos_5, edge_color='gray', width=2)
nx.draw_networkx_labels(G5, pos_5, font_size=11, font_weight='bold')

plt.title("School Friendship Network Visualization")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd

# 1. Create graph representing tower connections
G6 = nx.Graph()
tower_links = [
    ("T1", "T2"), ("T1", "T3"),
    ("T2", "T4"), ("T3", "T4"),
    ("T4", "T5"), ("T5", "T6"),
    ("T3", "T6"), ("T6", "T7"),
    ("T2", "T7")
]
G6.add_edges_from(tower_links)

# 3. Total number of nodes and edges
print(f"Total Nodes: {G6.number_of_nodes()}")
print(f"Total Edges: {G6.number_of_edges()}")

# 4. Degree of each node and highest degree tower(s)
degrees_6 = dict(G6.degree())
max_degree_6 = max(degrees_6.values())
top_towers = [node for node, deg in degrees_6.items() if deg == max_degree_6]
print("\nNode Degrees:", degrees_6)
print(f"Tower(s) with highest degree ({max_degree_6}): {', '.join(top_towers)}")

# 5. Shortest path and length between T1 and T7
shortest_path_6 = nx.shortest_path(G6, source="T1", target="T7")
path_length_6 = nx.shortest_path_length(G6, source="T1", target="T7")
print(f"\nShortest path between T1 and T7: {' -> '.join(shortest_path_6)}")
print(f"Shortest path length: {path_length_6}")

# 6. Centrality Measures
deg_cent = nx.degree_centrality(G6)
between_cent = nx.betweenness_centrality(G6)
close_cent = nx.closeness_centrality(G6)

centrality_df = pd.DataFrame({
    'Degree Centrality': deg_cent,
    'Betweenness Centrality': between_cent,
    'Closeness Centrality': close_cent
})
print("\nCentrality Measures:")
print(centrality_df.round(3))

# 7. Check connectivity and connected components
is_connected = nx.is_connected(G6)
num_components = nx.number_connected_components(G6)
print(f"\nIs network fully connected?: {is_connected}")
print(f"Number of connected components: {num_components}")

# 8. Highlight tower with highest betweenness centrality
highest_bw_tower = max(between_cent, key=between_cent.get)

node_colors = ['crimson' if node == highest_bw_tower else 'skyblue' for node in G6.nodes()]
node_sizes = [1600 if node == highest_bw_tower else 900 for node in G6.nodes()]

pos_6 = nx.spring_layout(G6, seed=42)
plt.figure(figsize=(9, 7))

nx.draw_networkx_nodes(G6, pos_6, node_color=node_colors, node_size=node_sizes)
nx.draw_networkx_edges(G6, pos_6, edge_color='gray', width=2)
nx.draw_networkx_labels(G6, pos_6, font_size=10, font_weight='bold')

plt.title(f"Telecom Network (Highlighted Highest Betweenness: {highest_bw_tower})")
plt.axis('off')
plt.tight_layout()
plt.show()
```
