```python
import pandas as pd

# Define the common labels
cols = ["C1", "C2", "C3"]
rows = ["R1", "R2", "R3"]

# 1. Create DataFrame A
data_A = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
df_A = pd.DataFrame(data_A, columns=cols, index=rows)

# 2. Create DataFrame B
data_B = [[9, 8, 7], [6, 5, 4], [3, 2, 1]]
df_B = pd.DataFrame(data_B, columns=cols, index=rows)

# Perform Operations
print("1. Addition (A + B):\n", df_A + df_B)
print("\n2. Subtraction (A - B):\n", df_A - df_B)
print("\n3. Element-wise Multiplication (A * B):\n", df_A * df_B)

# 4. Matrix Multiplication (Fixing Alignment)
# We convert to numpy values or ensure index of B matches columns of A
df_dot = pd.DataFrame(
    df_A.values @ df_B.values,
    index=rows,
    columns=cols
)
print("\n4. Matrix Multiplication (Fixed using @ operator):\n", df_dot)
```

```python
import pandas as pd

# Create the DataFrame
data = {
    'Student': ['Amit', 'Bhavna', 'Chirag', 'Divya', 'Esha'],
    'Maths': [85, 70, 95, 60, 88],
    'Science': [90, 65, 91, 75, 84],
    'English': [88, 72, 93, 70, 86],
    'History': [92, 68, 96, 80, 90]
}

df = pd.DataFrame(data)
subjects = ['Maths', 'Science', 'English', 'History']

print("--- Initial DataFrame ---")
print(df)
print("\n" + "="*40 + "\n")

# 1. Average marks of each student
df['Average'] = df[subjects].mean(axis=1)
print("1. Average Marks of Each Student:")
print(df[['Student', 'Average']].to_string(index=False))
print("\n" + "="*40 + "\n")

# 2. Average marks of each subject
subject_avg = df[subjects].mean()
print("2. Average Marks of Each Subject:")
print(subject_avg.to_string())
print("\n" + "="*40 + "\n")

# 3. Highest and lowest marks (Overall & Per Subject)
overall_max = df[subjects].max().max()
overall_min = df[subjects].min().min()
print("3. Highest and Lowest Marks:")
print(f"   - Overall Highest Mark: {overall_max}")
print(f"   - Overall Lowest Mark:  {overall_min}\n")

print("   Highest Mark Per Subject:")
print(df[subjects].max().to_string())
print("\n   Lowest Mark Per Subject:")
print(df[subjects].min().to_string())
print("\n" + "="*40 + "\n")

# 4. Student having the highest total marks
df['Total'] = df[subjects].sum(axis=1)
top_student_idx = df['Total'].idxmax()
top_student = df.loc[top_student_idx, 'Student']
top_score = df.loc[top_student_idx, 'Total']

print(f"4. Student with Highest Total Marks:\n   {top_student} with {top_score} total marks")
print("\n" + "="*40 + "\n")

# 5. Sort the DataFrame by total marks in descending order
df_sorted = df.sort_values(by='Total', ascending=False)
print("5. DataFrame Sorted by Total Marks (Descending):")
print(df_sorted[['Student', 'Maths', 'Science', 'English', 'History', 'Total', 'Average']].to_string(index=False))
```

```python
import pandas as pd
import numpy as np

# 0. Construct the DataFrame
data = {
    'Day': ['Day1', 'Day2', 'Day3', 'Day4', 'Day5', 'Day6', 'Day7'],
    'Temperature': [30.0, np.nan, 32.0, 29.0, np.nan, 35.0, 31.0]
}

df = pd.DataFrame(data)
print("--- Initial DataFrame ---")
print(df)
print("\n" + "="*45 + "\n")

# 1. Identify missing values using isnull()
print("1. Identifying Missing Values (isnull):")
print(df.isnull())
print("\nMissing values per column:")
print(df.isnull().sum())
print("\n" + "="*45 + "\n")

# 2. Fill missing values with the mean temperature
mean_temp = df['Temperature'].mean()
df_filled_mean = df.copy()
df_filled_mean['Temperature'] = df_filled_mean['Temperature'].fillna(mean_temp)

print(f"2. Fill Missing Values with Mean (Mean = {mean_temp:.1f}°C):")
print(df_filled_mean)
print("\n" + "="*45 + "\n")

# 3. Fill missing values using forward fill (ffill)
df_ffill = df.copy()
df_ffill['Temperature'] = df_ffill['Temperature'].ffill()

print("3. Fill Missing Values using Forward Fill (ffill):")
print(df_ffill)
print("\n" + "="*45 + "\n")

# 4. Drop rows containing missing values
df_dropped = df.dropna()

print("4. Drop Rows with Missing Values:")
print(df_dropped)
print("\n" + "="*45 + "\n")

# 5. Calculate average temperature after cleaning the data
avg_dropped = df_dropped['Temperature'].mean()
avg_filled = df_filled_mean['Temperature'].mean()
avg_ffill = df_ffill['Temperature'].mean()

print("5. Average Temperature After Cleaning:")
print(f"   - After dropping NaNs (5 valid days):     {avg_dropped:.2f}°C")
print(f"   - After mean-imputation (7 days):          {avg_filled:.2f}°C")
print(f"   - After forward-fill imputation (7 days):  {avg_ffill:.2f}°C")
```

```python
import pandas as pd

# Create the initial DataFrame
data = {
    'Product': ['Product A', 'Product B', 'Product C', 'Product D'],
    'Jan': [120, 100, 150, 90],
    'Feb': [130, 110, 145, 100],
    'Mar': [140, 120, 160, 110],
    'Apr': [150, 130, 170, 120],
    'May': [160, 140, 180, 130],
    'Jun': [170, 150, 190, 140]
}

df = pd.DataFrame(data)

# Reshape into long format for groupby and pivot operations
df_long = df.melt(id_vars=['Product'], var_name='Month', value_name='Sales')

print("--- Original DataFrame ---")
print(df)
print("\n" + "="*50 + "\n")

# 1. Total sales of each product using groupby / sum()
total_sales = df_long.groupby('Product')['Sales'].sum()
print("1. Total Sales of Each Product (groupby/sum):")
print(total_sales.to_string())
print("\n" + "="*50 + "\n")

# 2. Average monthly sales of each product
avg_sales = df_long.groupby('Product')['Sales'].mean()
print("2. Average Monthly Sales of Each Product:")
print(avg_sales.round(2).to_string())
print("\n" + "="*50 + "\n")

# 3. Best-selling product overall
best_product = total_sales.idxmax()
best_product_sales = total_sales.max()
print(f"3. Best-selling Product Overall:\n   {best_product} ({best_product_sales} units)")
print("\n" + "="*50 + "\n")

# 4. Month with the highest overall sales
monthly_sales = df_long.groupby('Month')['Sales'].sum()
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()
print(f"4. Month with Highest Overall Sales:\n   {best_month} ({best_month_sales} units)")
print("\n" + "="*50 + "\n")

# 5. Standard deviation of sales for each product
std_sales = df_long.groupby('Product')['Sales'].std()
print("5. Standard Deviation of Sales for Each Product:")
print(std_sales.round(2).to_string())
print("\n" + "="*50 + "\n")

# 6. Pivot table showing total sales per product per quarter
# Map months to quarters (Q1: Jan–Mar, Q2: Apr–Jun)
q1_months = ['Jan', 'Feb', 'Mar']
df_long['Quarter'] = df_long['Month'].apply(lambda m: 'Q1' if m in q1_months else 'Q2')

# Create pivot table
quarter_pivot = pd.pivot_table(
    df_long,
    values='Sales',
    index='Product',
    columns='Quarter',
    aggfunc='sum'
)

print("6. Pivot Table (Total Sales per Product per Quarter):")
print(quarter_pivot)
```
