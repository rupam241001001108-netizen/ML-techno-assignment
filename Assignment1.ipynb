```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
second_row = matrix[1]
print("Second row:", second_row)
third_column = [row[2] for row in matrix]
print("Third column:", third_column)
total_sum = sum(sum(row) for row in matrix)
print("Sum of all elements:", total_sum)
```

```python

students_marks = [
    ["Alice", [85, 90, 88]],
    ["Bob", [92, 95, 94]],
    ["Charlie", [78, 82, 80]],
    ["Diana", [88, 85, 91]]
]

averages = {name: sum(marks) / len(marks) for name, marks in students_marks}
for name, avg in averages.items():
    print(f"{name}: {avg:.2f}")
top_student = max(averages, key=averages.get)
print(f"\nTop student: {top_student} with an average of {averages[top_student]:.2f}")
```

```python

student = {
    "Name": "Alice",
    "Roll Number": 101,
    "Department": "Computer Science",
    "CGPA": 8.5
}
print("--- Initial Student Details ---")
for key, value in student.items():
    print(f"{key}: {value}")

student["CGPA"] = 9.1
print("\n--- Updated Student Details ---")
print(f"Updated CGPA: {student['CGPA']}")
```

```python
company = {
    "HR": ["Alice", "Bob"],
    "Engineering": ["Charlie", "David", "Eve"],
    "Sales": ["Frank"]
}

target_dept = "Engineering"
print(f"Employees in {target_dept}: {company[target_dept]}")

print("Employee count per department:")
for dept, employees in company.items():
    print(f"{dept}: {len(employees)}")
```

```python
student_data = {
    "Name": "Emma",
    "Marks": [85, 92, 78, 90, 88]
}

marks = student_data["Marks"]
average = sum(marks) / len(marks)
highest = max(marks)
lowest = min(marks)

print(f"Student: {student_data['Name']}")
print(f"Average: {average}, Highest: {highest}, Lowest: {lowest}")
```

```python
students = [
    {"Name": "Alice", "Marks": 85},
    {"Name": "Bob", "Marks": 92},
    {"Name": "Charlie", "Marks": 88}
]

highest_mark = 0
topper = ""

print("Student Names:")
for student in students:
    print(student["Name"])
    if student["Marks"] > highest_mark:
        highest_mark = student["Marks"]
        topper = student["Name"]

print(f"\nThe topper is {topper} with {highest_mark} marks.")
```

```python
books = [
    {"Title": "Book A", "Price": 450},
    {"Title": "Book B", "Price": 700},
    {"Title": "Book C", "Price": 300}
]

total_cost = 0
most_expensive = max(books, key=lambda x: x["Price"])

for book in books:
    total_cost += book["Price"]

print(f"Most expensive book: {most_expensive['Title']} (${most_expensive['Price']})")
print(f"Total cost of all books: ${total_cost}")
```

```python
import array

arr = array.array('i', [10, 20, 5, 40, 50, 15, 70, 80, 25, 30])

maximum = max(arr)
minimum = min(arr)
average = sum(arr) / len(arr)

print(f"Array: {arr.tolist()}")
print(f"Max: {maximum}, Min: {minimum}, Average: {average}")
```

```python
import array

arr = array.array('i', [1, 2, 3, 4, 5])

# Insert new element (e.g., 99 at index 2)
arr.insert(2, 99)
print("After insertion:", arr.tolist())

# Delete an existing element (e.g., 3)
arr.remove(3)
print("After deletion:", arr.tolist())

# Display in reverse order
arr.reverse()
print("Reversed array:", arr.tolist())
```

```python
student_tuple = ("Liam", 20, "Electrical", 8.2)
print("Original Tuple:", student_tuple)

# Convert to list
student_list = list(student_tuple)

# Update CGPA (index 3)
student_list[3] = 8.9

# Convert back to tuple
student_tuple = tuple(student_list)
print("Updated Tuple:", student_tuple)
```

```python

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
print("Difference (set1 - set2):", set1.difference(set2))
print("Symmetric Difference:", set1.symmetric_difference(set2))
```
